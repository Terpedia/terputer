# GSE162256: paclitaxel response in A2780 single cells

**Question.** Do predeclared paclitaxel response markers from the HCC1143 example show a similar direction in an independent A2780 cell context?

**Input.** GEO GSE162256 processed `Evalues.txt.gz`, pinned by SHA-256 in `analysis.py`. The script downloads and verifies the public input. Its `States` row is excluded from the numeric matrix.

**Design.** Compare 46 control cells with 56 paclitaxel 16 h cells and 53 paclitaxel 48 h cells. The 13 marker genes were predeclared from the HCC1143 analysis. Report medians in the source's reported units and positive-value detection fractions. Compare marker differences with all features detected in at least 20% of control cells. A cell is not a biological replicate. No p-values or cross-study fold changes are computed.

**Result.** All six cell-cycle markers have lower A2780 median values at both paclitaxel times than control. The feature-wide background also shifts lower, so marker direction alone is not cell-cycle-specific. Interferon genes are mixed; IFIT3, MX1, and OAS1 are absent from this processed matrix. HCC1143 showed lower MKI67/TOP2A/CENPF but higher CCNB1/UBE2C marker ratios, illustrating context or assay dependence. The source study emphasizes spike-in normalization and global RNA change; these reported E-values should not be treated as raw counts or directly comparable CPM.

**Limit.** Both datasets lack independent exposed/control culture replicates for this comparison. The cell lines, exposure times and doses, platform, and normalization differ. These patterns are descriptive and do not establish a reproducible pathway or a paclitaxel target mechanism.

Sources: [GEO GSE162256](https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE162256), [published study](https://pmc.ncbi.nlm.nih.gov/articles/PMC8210883/), [HCC1143 GSE266934](https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE266934).
